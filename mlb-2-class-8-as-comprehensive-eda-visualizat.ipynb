```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

```python
# Pie chart (Survived distribution)
titanic['Survived'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(5,5))
plt.title('Survival Distribution')
plt.ylabel('')
plt.show()
```

# 2. Count plot

```python
# Count plot
sns.countplot(data=titanic, x='Survived')
plt.title('Count of Survived Passengers')
plt.show()
```

# 3. Kde + Rug plot

```python
# KDE + Rug plot
sns.kdeplot(data=titanic, x='Age', fill=True)
sns.rugplot(data=titanic, x='Age')
plt.title('Age Distribution KDE + Rug')
plt.show()
```

# 4. Violin plot

```python
# Violin plot
sns.violinplot(data=titanic, x='Pclass', y='Age')
plt.title('Age Distribution by Passenger Class')
plt.show()
```

# 5. Box plot

```python
# Box plot
sns.boxplot(data=titanic, x='Pclass', y='Fare')
plt.title('Fare Distribution by Class')
plt.show()
```

# 6. Boxen plot

```python
# Boxen plot
sns.boxenplot(data=titanic, x='Pclass', y='Age')
plt.title('Age Distribution Boxen Plot')
plt.show()
```

# 7. Histogram

```python
# Histogram
sns.histplot(data=titanic, x='Age', bins=30, kde=True)
plt.title('Age Histogram')
plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
# Scatter plot
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex')
plt.title('Total Bill vs Tip')
plt.show()
```

# 2. Line plot

```python
# Line plot
sns.lineplot(data=tips, x='size', y='tip')
plt.title('Tip Trend by Group Size')
plt.show()
```

# 3. Joint plot

```python
# Joint plot
sns.jointplot(data=tips, x='total_bill', y='tip', kind='scatter')
plt.show()
```

# 4. Hexbin plot

```python
# Hexbin plot
plt.hexbin(tips['total_bill'], tips['tip'], gridsize=20)
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Hexbin Plot')
plt.colorbar()
plt.show()
```

# 5. 2D kde plot

```python
# 2D KDE plot
sns.kdeplot(data=tips, x='total_bill', y='tip', fill=True)
plt.title('2D KDE Plot')
plt.show()
```

# 6. Box plot

```python
# Box plot
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill by Day')
plt.show()
```

# 7. Violin plot

```python
# Violin plot
sns.violinplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill Distribution by Day')
plt.show()
```

# 8. Swarm plot

```python
# Swarm plot
sns.swarmplot(data=tips, x='day', y='tip')
plt.title('Tip Distribution by Day')
plt.show()
```

# 9. Count plot

```python
# Count plot
sns.countplot(data=tips, x='day')
plt.title('Number of Customers by Day')
plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
sns.scatterplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.title('Sepal Length vs Petal Length')
plt.show()
```

# 2. Pair plot

```python
sns.pairplot(
    iris,
    hue='species'
)

plt.show()
```

# 3. Heatmap

```python
# Heatmap
numeric_iris = iris.select_dtypes(include=np.number)
sns.heatmap(numeric_iris.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

# 4. FacetGrid plot

```python
# FacetGrid plot

g = sns.FacetGrid(iris, col='species')

g.map_dataframe(
    sns.scatterplot,
    x='sepal_length',
    y='sepal_width'
)

plt.show()
```

# 5. PairGrid

```python
# PairGrid

g = sns.PairGrid(iris, hue='species')

g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)

g.add_legend()

plt.show()
```

# 6. Joint plot

```python
# Joint plot

sns.jointplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    kind='reg'
)

plt.show()
```

# 7. Boxen plot

```python
# Boxen plot

sns.boxenplot(
    data=iris,
    x='species',
    y='petal_length'
)

plt.title('Petal Length by Species')
plt.show()
```
