# 6. Apply below mathematical operations on a series

```python
# Create a series

series = pd.Series([10, 20, 30, 40, 50, 20, 30])

# count
print("Count:")
print(series.count())

# sum
print("\nSum:")
print(series.sum())

# product
print("\nProduct:")
print(series.product())

# min
print("\nMinimum:")
print(series.min())

# max
print("\nMaximum:")
print(series.max())

# mean
print("\nMean:")
print(series.mean())

# median
print("\nMedian:")
print(series.median())

# mode
print("\nMode:")
print(series.mode())

# describe
print("\nDescribe:")
print(series.describe())

# standard deviation
print("\nStandard Deviation:")
print(series.std())

# variance
print("\nVariance:")
print(series.var())
```

# Merging and joining

```python
students = pd.DataFrame({
    "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110,
                   111, 112, 113, 114, 115, 116, 117, 118, 119, 120],
    "name": ["Rahim", "Karim", "Hasan", "Sakib", "Nabil",
             "Tanvir", "Fahim", "Rafi", "Imran", "Shakil",
             "Arif", "Jahid", "Mehedi", "Siam", "Tareq",
             "Rakib", "Adnan", "Nayeem", "Foysal", "Rony"],
    "department": ["CSE", "EEE", "CSE", "BBA", "CSE",
                   "EEE", "CSE", "BBA", "CSE", "EEE",
                   "CSE", "BBA", "EEE", "CSE", "BBA",
                   "CSE", "EEE", "CSE", "BBA", "EEE"]
})

students
```

```python
results = pd.DataFrame({
    "student_id": [101, 102, 103, 105, 106, 108, 109, 111, 112, 114,
                   121, 122, 123, 124, 125, 126, 127, 128, 129, 130],
    "math": [85, 72, 91, 68, 77, 88, 74, 95, 81, 69,
             78, 84, 92, 73, 89, 67, 80, 94, 76, 83],
    "python": [90, 75, 88, 72, 80, 92, 79, 96, 85, 71,
               82, 87, 95, 76, 91, 70, 84, 93, 78, 86]
})

results
```

```python
departments = pd.DataFrame({
    "dept_id": [1, 2, 3, 4],
    "department": [
        "CSE",
        "EEE",
        "BBA",
        "Civil"
    ],
    "faculty": [
        "Engineering",
        "Engineering",
        "Business",
        "Engineering"
    ]
})

departments
```

# 1. Create a dataframe contains students name,faculty and number on math and python

```python
# Merge students and results using student_id

student_result = pd.merge(students, results, on="student_id")

# Merge with departments using department column

final_df = pd.merge(student_result, departments, on="department")

# Select required columns

final_df[["name", "faculty", "math", "python"]]
```

# 1. convert time into datetime

```python
# Convert date column into datetime format

climate["date"] = pd.to_datetime(climate["date"])

climate.head()
```

# 2. set date as index

```python
# Set date column as index

climate = climate.set_index("date")

climate.head()
```

# 3. Select row of 13 january, 2015

```python
# Select 13 January 2015

climate.loc["2015-01-13"]
```

# 4. Select rows contains in april 2017

```python
# Check rows in April 2017

climate.loc["2017-04"]
```

# 5. Select rows from 11 september, 2014 to 25 september, 2014

```python
# Select rows between two dates

climate.loc["2014-09-11":"2014-09-25"]
```

# 6. Calculate month-wise average windspeed

```python
# Month-wise average wind speed

climate["month"] = climate.index.month

climate.groupby("month")["wind_speed"].mean()
```

# 7. Calculate month-wise highest meantemp

```python
# Month-wise highest mean temperature

climate["month"] = climate.index.month

climate.groupby("month")["meantemp"].max()
```

# 8. Year-wise maximum, minimum and average humidity

```python
# Extract year from date index

climate["year"] = climate.index.year

# Year-wise humidity statistics

climate.groupby("year")["humidity"].agg(
    ["max", "min", "mean"]
)
```

# 9. A new column contains previous 10 days average meantemp

```python
# Create new column with previous 10 days average temperature

climate["previous_10_days_avg_meantemp"] = (
    climate["meantemp"].rolling(window=10).mean()
)

climate.head(15)
```
